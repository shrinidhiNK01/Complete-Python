```python
class Movies:
    actor="Shrinidhi"
    actress="Shruthi"
    director="Lokesh Kanagaraj"
    Villain="Vijay Sethupathi"
    budget="20cr"
    collection="100cr"
print(Movies.actor)
Movies.actor="Vikram"
print(Movies.actor)
print(Movies.actress)
print(Movies.director)
print(Movies.Villain)
print(Movies.budget)
print(Movies.collection)
print(Movies.director)
Movies.sideChick="Samantha"
print(Movies.sideChick)
```

```python
class KGF:
    actor="Yash"
    actress="Srinidhi"
    director="Prashanth Neel"
    Villain="Vijay Sethupathi"
    budget="250cr"
    collection="1000cr"
    def Movie(self):   #Self:- Reference variable that refers to the current instance of the class, used to access variables that belongs to the class
        print(f"KGF is a great movie with {self.collection} is the collection of the movie and {self.budget} is the budget of the movie")

k=KGF()
k.Movie()
```

```python
#Constructor:- A special method that is called when an object is instantiated, used to initialize the variables of the class
#Constructor will get executed automatically when an object is created, we don't need to call the constructor explicitly
class KGF:
    def __init__(self,actor,actress,director,Villain,budget,collection):
        print("This is the constructor of the class KGF")
        self.actor=actor
        self.actress=actress
        self.director=director
        self.Villain=Villain
        self.budget=budget
        self.collection=collection
    def Movie(self):   #Self:- Reference variable that refers to the current instance of the class, used to access variables that belongs to the class
        print(f"KGF is a great movie with {self.collection} is the collection of the movie and {self.budget} is the budget of the movie")
    
k=KGF("Yash","Srinidhi","Prashanth Neel","Vijay Sethupathi","250cr","1000cr")
k=KGF("Darshan","Samantha","Rajkumar","Naveen Kasturi","250cr","300cr")
# k=KGF("Darshan","Samantha","Rajkumar","Naveen Kasturi","250cr","300cr")
k.Movie()
k.Movie()

# No of objects created in the class = no of times the constructor is called
```

```python
# AsyncIO :- allows high performance concurrent code using the async/await syntax, it is used to write asynchronous code that can run concurrently without blocking the main thread, it is used to improve the performance of I/O bound tasks such as network requests, file handling, etc.
import asyncio

async def func1():
    await asyncio.sleep(15)
    print("I am god")

async def func2():
    await asyncio.sleep(10)
    print("I am second god")

async def func3():
    await asyncio.sleep(5)
    print("I am third god")

async def main():
    await asyncio.gather(func1(),func2(),func3())

await main()
```
